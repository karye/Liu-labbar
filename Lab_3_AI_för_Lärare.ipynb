### Uppgift 3 – Ljud till text: transkribera nyheter

**Scenario:** Du ska inleda lektionen med att arbeta med en nyhet där ni tillsammans lyssnar på ett nyhetsinslag. Du märker dock att vissa elever inte riktigt lyssnar aktivt eller snabbt glömmer innehållet. Istället för att spela upp klippet flera gånger och gå igenom alla frågor muntligt, vill du ha tillgång till innehållet som text.

**Verktyg:** Whisper (tal-till-text)

Idag finns det verktyg som kan omvandla tal till text och även översätta språk. Ett sådant verktyg är Whisper. Kod för att köra Whisper återfinns i kodcellerna nedan.

#### Steg 1 – Transkribera ett nyhetsinslag

Här laddar vi ner ett nyhetsinslag och skapar en textversion av det. Vill du använda ett annat inslag kan du byta ut länken i variabeln URL. Som standard används denna länk:

> <https://www.sverigesradio.se/artikel/krigsbloggare-rasar-nar-telegram-blockeras-i-ryssland>

Verktyget transkriberar ljudet till text och kan även översätta det.

> **OBS:** Transkriberingen kan ta lite tid.


```python
# --- Installs ---
!pip install -q yt-dlp openai-whisper fpdf2

# --- Imports ---
import whisper
import subprocess
import os
import textwrap
import requests
import re
from fpdf import FPDF
from pathlib import Path
from datetime import datetime
```

```python
URL = "https://www.sverigesradio.se/artikel/krigsbloggare-rasar-nar-telegram-blockeras-i-ryssland"
OUTPUT_DIR = "/content/sr_audio"
WHISPER_MODEL = "medium"  # "small" is faster, "medium"/"large" better for Swedish

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ── 1. Scrape the audio URL from the article page ──────────────────────
page = requests.get(URL).text

audio_match = re.search(r'https?://www\.sverigesradio\.se/topsy/ljudfil/\d+', page)
if not audio_match:
    audio_match = re.search(r'"audioId\\?":\s*(\d+)', page)
    if not audio_match:
        raise RuntimeError("Could not find any audio on this article page.")
    audio_url = f"https://www.sverigesradio.se/topsy/ljudfil/{audio_match.group(1)}"
else:
    audio_url = audio_match.group(0).replace("\\", "")

print(f"Found audio URL: {audio_url}")

# ── 2. Download the MP3 (follow redirects to CDN) ──────────────────────
audio_path = os.path.join(OUTPUT_DIR, "audio.mp3")
resp = requests.get(audio_url, allow_redirects=True)
resp.raise_for_status()
with open(audio_path, "wb") as f:
    f.write(resp.content)
print(f"Audio downloaded: {audio_path} ({len(resp.content) / 1024:.0f} KB)")

# ── 3. Transcribe with Whisper ──────────────────────────────────────────
model = whisper.load_model(WHISPER_MODEL)
transcription = model.transcribe(audio_path, language="sv")
text = transcription["text"].strip()
print(f"\n--- Transcription ({len(text)} chars) ---\n{text}\n")

# ── 4. Build PDF ────────────────────────────────────────────────────────
FONT_PATH = "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf"
if not os.path.exists(FONT_PATH):
    FONT_PATH = "/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf"

pdf = FPDF()
pdf.add_page()
pdf.add_font("UniFont", "", FONT_PATH, uni=True)

pdf.set_font("UniFont", size=16)
pdf.cell(0, 12, "Sveriges Radio – Transcription", ln=True, align="C")

pdf.set_font("UniFont", size=9)
pdf.set_text_color(120, 120, 120)
pdf.cell(0, 8, f"Source: {URL}", ln=True, align="C")
pdf.cell(0, 6, f"Generated: {datetime.now().strftime('%Y-%m-%d %H:%M')}", ln=True, align="C")
pdf.ln(6)

pdf.set_text_color(0, 0, 0)
pdf.set_font("UniFont", size=11)

for paragraph in text.split("\n"):
    paragraph = paragraph.strip()
    if paragraph:
        pdf.multi_cell(0, 7, paragraph)
        pdf.ln(3)

pdf_path = os.path.join(OUTPUT_DIR, "transcription.pdf")
pdf.output(pdf_path)
print(f"PDF saved to: {pdf_path}")

# Auto-download in Colab
try:
    from google.colab import files
    files.download(pdf_path)
except ImportError:
    pass
```
