## 📖 Del 1 – Vad är Machine Learning?

Föreställ dig att din e-post automatiskt sorterar bort **spam**. Hur vet datorn att ett mail är spam?

Den har **lärt sig** från tusentals exempel!

```
Spam-exempel:         Inte spam:
"Vinn 1 miljon nu!"  "Hej, kan vi ses imorgon?"
"Klicka här GRATIS"  "Mötet är kl 14:00"
"SPECIELLT ERBJUDANDE" "Glöm inte ta med lunch"
```

Datorn analyserar **mönster** i texten och lär sig skilja spam från vanliga mail.

**Det är precis vad Machine Learning (ML) är!**

> 🤖 *Machine Learning = Datorer lär sig från EXEMPEL, istället för att vi programmerar varje regel*

### Tre typer av Machine Learning:

| Typ | Förklaring | Exempel |
|-----|-----------|---------|
| **Supervised Learning** | Vi ger modellen märkta exempel (input + rätt svar) | Spam-filter, blomklass |
| **Unsupervised Learning** | Modellen hittar mönster själv, utan märkta svar | Gruppera kunder |
| **Reinforcement Learning** | Modellen lär sig genom att prova och få belöning/straff | Spela schack |

## 📊 Del 4 – Dataset-konceptet

### Features och Labels

I ML pratar vi om:

| Term | Förklaring | Iris-exempel |
|------|-----------|-------------|
| **Feature** | En egenskap hos datapunkten | Sepal length = 5.1 cm |
| **Label** | Det rätta svaret | "Setosa" |
| **Sample/rad** | En datapunkt | En specifik blomma |
| **Dataset** | Alla datapunkter tillsammans | 150 blommor |

### Träningsdata vs Testdata

Vi delar alltid upp vårt dataset i **två delar**:

```
HELA DATASETET (150 blommor)
┌──────────────────────────────────────────┐
│  TRÄNINGSDATA (120 blommor, 80%)         │
│  ← Modellen lär sig från dessa           │
├──────────────────────────────────────────┤
│  TESTDATA (30 blommor, 20%)              │
│  ← Vi testar hur bra modellen är         │
└──────────────────────────────────────────┘
```

**Varför dela upp?**  
Vi vill inte att modellen "fuskar" genom att memorera svaren.  
Testdata är svar modellen **aldrig sett** – det är ett riktigt test!

## 🎮 Del 5 – Interaktiv demo: Se ett dataset!

Nu ska du få **utforska** hur ett dataset ser ut och hur klassificering fungerar!

Kör cellen nedan och använd **slidern** för att byta mellan blommor i Iris-datasetet.

````python
# ============================================================
# INTERAKTIV DEMO: Utforska Iris-datasetet
# ============================================================

# Ladda Iris-datasetet
iris = load_iris()
feature_names = ['Sepal längd (cm)', 'Sepal bredd (cm)',
                  'Petal längd (cm)', 'Petal bredd (cm)']
# Vi skapar en ren lista för titlar utan emojis för att undvika font-fel
class_names = ['Setosa 🌸', 'Versicolor 🌺', 'Virginica 🌻']
class_names_clean = ['Setosa', 'Versicolor', 'Virginica']
colors = ['#FF9AA2', '#FFB347', '#87CEEB']

# Skapa en DataFrame för enklare hantering
df = pd.DataFrame(iris.data, columns=feature_names)
df['Art'] = [class_names[t] for t in iris.target]

# ── Interaktiv widget ──────────────────────────────────────
output = widgets.Output()

def visa_blomma(blomma_nr):
    with output:
        clear_output(wait=True)
        art_index = iris.target[blomma_nr]

        print(f"🌸 Blomma nummer {blomma_nr + 1} av 150")
        print("=" * 40)
        print(f"  Art: {class_names[art_index]}")
        print()
        print("  Mätningar (Features):")
        for fn, val in zip(feature_names, iris.data[blomma_nr]):
            bar = '█' * int(val * 3)
            print(f"  {fn:22s}: {val:.1f} {bar}")
        print()

        # Rita ett stapeldiagram för denna blomma
        fig, ax = plt.subplots(figsize=(7, 3))
        bars = ax.bar(feature_names, iris.data[blomma_nr],
                      color=colors[art_index], edgecolor='black', linewidth=0.5)

        # Vi använder den rena texten här för att undvika varningar
        ax.set_title(f'Blomma {blomma_nr + 1}: Iris {class_names_clean[art_index]}',
                     fontsize=13, fontweight='bold')
        ax.set_ylabel('Värde (cm)')
        ax.set_ylim(0, 8)
        for bar, val in zip(bars, iris.data[blomma_nr]):
            ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
                    f'{val:.1f}', ha='center', va='bottom', fontsize=10)
        plt.xticks(rotation=20, ha='right')
        plt.tight_layout()
        plt.show()

blomma_slider = widgets.IntSlider(
    value=0, min=0, max=149, step=1,
    description='Blomma nr:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='500px')
)

# Koppla ihop slidern med funktionen och visa allt
interactive_plot = widgets.interactive(visa_blomma, blomma_nr=blomma_slider)
display(widgets.VBox([interactive_plot, output]))
````

### 🔍 Interaktiv Scatterplot – Se hur arterna skiljer sig!

Kör cellen nedan och **välj vilka features** du vill jämföra med dropdown-menyer.  
Kan du hitta features som tydligt separerar de tre arterna?

````python
# ============================================================
# INTERAKTIV SCATTERPLOT
# ============================================================

output2 = widgets.Output()

def rita_scatter(x_feature, y_feature):
    with output2:
        clear_output(wait=True)

        xi = feature_names.index(x_feature)
        yi = feature_names.index(y_feature)

        fig, ax = plt.subplots(figsize=(8, 5))

        # Vi använder class_names_clean här för att slippa font-varningar i legenden
        for i, (art_name, color) in enumerate(zip(class_names_clean, colors)):
            mask = iris.target == i
            ax.scatter(iris.data[mask, xi], iris.data[mask, yi],
                       c=color, label=art_name, s=60, edgecolors='black', linewidth=0.5)

        ax.set_xlabel(x_feature, fontsize=11)
        ax.set_ylabel(y_feature, fontsize=11)
        ax.set_title(f'Iris-datasetet: {x_feature} vs {y_feature}',
                     fontsize=13, fontweight='bold')
        ax.legend(title='Art', fontsize=10)
        ax.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()

        # Tips
        if xi == yi:
            print("⚠️ Välj OLIKA features på x- och y-axeln!")
        else:
            setosa_x = iris.data[iris.target==0, xi]
            others_x = iris.data[iris.target!=0, xi]
            overlap = min(setosa_x.max(), others_x.max()) - max(setosa_x.min(), others_x.min())
            if overlap < 0.5:
                print(f"✅ Bra val! {x_feature} skiljer Setosa tydligt från de andra!")
            else:
                print(f"💡 Prova Petal längd eller Petal bredd för bättre separation!")

x_dropdown = widgets.Dropdown(
    options=feature_names, value=feature_names[2],
    description='X-axel:', style={'description_width': 'initial'}
)
y_dropdown = widgets.Dropdown(
    options=feature_names, value=feature_names[3],
    description='Y-axel:', style={'description_width': 'initial'}
)

ui = widgets.VBox([
    widgets.HTML('<b>🎮 Välj features att jämföra:</b>'),
    widgets.HBox([x_dropdown, y_dropdown]),
    output2
])

def on_change(change):
    rita_scatter(x_dropdown.value, y_dropdown.value)

x_dropdown.observe(on_change, names='value')
y_dropdown.observe(on_change, names='value')

rita_scatter(feature_names[2], feature_names[3])  # Visa direkt
display(ui)
````
