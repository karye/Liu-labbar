## 📝 Del 1 – Liknelsen: Provet du inte fått se

Föreställ dig den här situationen:

**Scenario A – Dålig lärare:**  
Läraren delar ut **exakt samma frågor** på provet som på övningsuppgifterna.  
Alla elever får 100% – men visar det att de förstår ämnet? 🤔

**Scenario B – Rättvis lärare:**  
Läraren ger övningsuppgifter för att träna på.  
På provet dyker det upp **nya frågor** som testar om du *verkligen* förstår.

```
ÖVNINGSUPPGIFTER         PROVET
(tränar på dessa)        (nya frågor du aldrig sett)
──────────────────       ────────────────────────────
Fråga 1, 2, 3, ...  →   Fråga 7, 8, 9, ...
```

**Exakt samma logik gäller för AI!**

Om vi tränar AI:n på **all** data och sedan testar den på **samma** data –  
vet vi inte om den verkligen har lärt sig, eller bara "memorerat" svaren.

> 🎯 Lösning: Dela upp data i **Träningsdata (Training Data)** och **Testdata (Test Data)**.
> - Träningsdata = övningsuppgifterna AI:n får lära sig av  
> - Testdata = det hemliga "provet" som AI:n inte fick se under träningen

## ⚙️ Del 2 – Ladda data och verktyg

Vi börjar med att ladda Iris-datasetet (från Lektion 1) och de verktyg vi behöver:

````python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Ladda datasetet
iris = load_iris(as_frame=True)
X = iris.data       # Egenskaper (Features) – de 4 mätningarna
y = iris.target     # Målvariabel (Target) – vilken art blomman är

print(f"Totalt antal blommor: {len(X)}")
print("✅ Data laddad!")
````

### Vad är X och y?

I maskininlärning har vi en convention (ett vanligt sätt att namnge saker):

- **X** (stor bokstav) = **egenskaperna (Features)** – det AI:n ska titta på  
  *Exempel: kalkbladets längd, kronbladets bredd...*

- **y** (liten bokstav) = **svaret (Target)** – det AI:n ska gissa  
  *Exempel: 0=Setosa, 1=Versicolor, 2=Virginica*

Det är som att X är framsidan på flashcard och y är baksidan!

## ✂️ Del 3 – Dela upp datan

Nu använder vi funktionen **`train_test_split`** för att dela upp datan.  
Parametern `test_size=0.2` betyder att 20% av datan sparas till testning och 80% används för träning.

Det är som att ta ett kortlek och säga:  
*"De flesta kort (80%) är övningsuppgifter. De sista (20%) är provet."*

> **`random_state=42`** gör att uppdelningen blir **samma** varje gång du kör koden.  
> Utan detta skulle uppdelningen vara slumpmässig och olika varje körning.

````python
# Dela upp datan: 80% träning, 20% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% sparas till test
    random_state=42     # Gör att vi alltid får samma uppdelning
)

print(f"Träningsdata: {len(X_train)} blommor")
print(f"Testdata:     {len(X_test)} blommor")
````

### Vad fick vi?

Vi delade upp 150 blommor i:
- **120 blommor** i träningsdatan – AI:n lär sig av dessa
- **30 blommor** i testdatan – dessa ser AI:n inte förrän vi utvärderar den

```
150 blommor totalt
┌─────────────────────────────────────────────────────┐
│  TRÄNINGSDATA (80%)          │  TESTDATA (20%)       │
│  120 blommor                 │  30 blommor           │
│  AI:n lär sig av dessa       │  Det hemliga provet   │
└─────────────────────────────────────────────────────┘
```

Notera att vi fick **fyra variabler**:
- `X_train` = egenskaperna för träningsblommorna
- `y_train` = arterna för träningsblommorna (de rätta svaren AI:n tränar på)
- `X_test` = egenskaperna för testblommorna
- `y_test` = de rätta svaren för testblommorna (vi håller dessa hemliga för AI:n!)

## 🔍 Del 4 – Inspektera uppdelningen

Låt oss kolla hur många av varje art som hamnade i tränings- respektive testdata:

````python
artnamn = {0: 'Setosa', 1: 'Versicolor', 2: 'Virginica'}

print("=== TRÄNINGSDATA ===")
for art_nr, namn in artnamn.items():
    antal = (y_train == art_nr).sum()
    print(f"  {namn}: {antal} blommor")

print()
print("=== TESTDATA ===")
for art_nr, namn in artnamn.items():
    antal = (y_test == art_nr).sum()
    print(f"  {namn}: {antal} blommor")
````
