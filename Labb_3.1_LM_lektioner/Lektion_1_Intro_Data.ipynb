## 🌺 Del 2 – Iris-datasetet

I denna lektion ska vi använda ett av världens mest kända AI-dataset:  
**Iris-datasetet** (iris = ett slags blomma).

Datasetet innehåller **150 blommor** från tre olika arter:

| Art | Utseende |
|-----|----------|
| 🌸 **Iris Setosa** | Liten, kompakt blomma |
| 🌺 **Iris Versicolor** | Mellanstor blomma |
| 🌻 **Iris Virginica** | Stor, vacker blomma |

För varje blomma har en botaniker mätt **4 saker** (på engelska kallade **Features / Egenskaper**):

```
1. sepal length  – kalkbladets längd (cm)
2. sepal width   – kalkbladets bredd (cm)
3. petal length  – kronbladets längd (cm)
4. petal width   – kronbladets bredd (cm)
```

> **Kalkblad (sepal)** = det yttre gröna/gula bladet  
> **Kronblad (petal)** = det färgade bladet i mitten

Botanikern har också skrivit upp **vilken art** varje blomma tillhör.  
Det kallar vi **Målvariabel (Target)** – det är svaret vi vill att datorn ska kunna gissa!

## 📊 Del 4 – Ladda och titta på datan

Nu laddar vi Iris-datasetet och lägger det i en **tabell (DataFrame)**.  
En DataFrame är som ett Excel-ark fast inne i Python.

Kör cellen nedan för att ladda datan:

````python
# Ladda Iris-datasetet
iris = load_iris(as_frame=True)

# Skapa en tabell med alla egenskaper
df = iris.data

# Lägg till en kolumn som visar vilken art blomman är
df['art'] = iris.target

# Visa de 10 första raderna
df.head(10)
````

### Vad ser vi?

Varje **rad** i tabellen är en enskild blomma.  
Varje **kolumn** innehåller antingen:
- En **egenskap (Feature)** – något vi mätt (längd, bredd)
- **Målvariabeln (Target)** – svaret, dvs. vilken art blomman är

I kolumnen `art` betyder:
- `0` = Iris Setosa
- `1` = Iris Versicolor  
- `2` = Iris Virginica

Datorn förstår inte namn, bara siffror – det är därför vi använder 0, 1 och 2.

---
## 🔍 Del 5 – Gör tabellen mer läsbar

Vi kan byta ut siffrorna mot artnamnen så att det blir lättare att läsa.  
Vi kan också byta kolumnnamnen till svenska:

````python
# Byt kolumnnamnen till svenska
df.columns = [
    'kalkblad_längd_cm',
    'kalkblad_bredd_cm',
    'kronblad_längd_cm',
    'kronblad_bredd_cm',
    'art'
]

# Byt siffrorna mot riktiga artnamn
artnamn = {0: 'Setosa', 1: 'Versicolor', 2: 'Virginica'}
df['art'] = df['art'].map(artnamn)

# Visa tabellen
df.head(15)
````

---
## 🔎 Del 6 – Utforska datasetet

Hur stor är datasetet? Hur ser fördelningen av arter ut?  
Låt oss ta reda på det:

````python
# Hur många rader och kolumner finns det?
print(f"Antal blommor: {len(df)}")
print(f"Antal egenskaper per blomma: {len(df.columns) - 1}")
print()

# Hur många av varje art?
print("Fördelning av arter:")
print(df['art'].value_counts())
````

Vi kan också se statistik över mätningarna – t.ex. minsta/största värde och medelvärde:

````python
# Statistik för varje kolumn
df.describe().round(2)
````
